## Word2Vec

1. Word2Vec is a technique for natural language processing published by Google in 2013. 
2. Uses Neural Network model to learn word associations from a large corpus of text.
3. Once trained, such a model can detect synonymous words or suggest additional words for a partial sentence.
4. Word2Vec represents each distinct word with a particular list of numbers called a vector.
