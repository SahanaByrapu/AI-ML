```python
import nltk
nltk.download('punkt')
```

```python
corpus = """ Hello Welcome, to San Francisco.
Hope you have great adventure! to become AI Engineer at Bay Area. """
```

```python
corpus
```

```python
## Tokenization

## Sentence --> paragraphs

from nltk.tokenize import sent_tokenize
```

```python
documents=sent_tokenize(corpus)
```

```python
type(documents)
```

```python
for sentence in documents:
    print(sentence)
```
